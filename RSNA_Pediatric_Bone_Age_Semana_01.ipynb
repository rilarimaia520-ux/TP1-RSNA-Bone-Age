## RSNA Pediatric Bone Age

**Desafio:** RSNA 2017 Pediatric Bone Age  
**Tipo de tarefa:** Regressão  
**Área:** Análise de imagens médicas  
**Unidade de amostra:** Radiografia 2D da mão  
**Variável de saída:** Idade óssea, em meses

## 1. Formulação do problema

O desafio consiste em estimar a idade óssea pediátrica a partir de radiografias 2D da mão. Para cada exame, a entrada do modelo será composta pela imagem radiográfica e pela informação de sexo disponível no conjunto de dados. A saída esperada será a idade óssea estimada, expressa em meses. Dessa forma, o problema é formulado como uma tarefa de regressão, na qual cada radiografia corresponde a uma unidade de amostra.

## 2. Referências e fichamentos

Nesta seção são apresentados os fichamentos iniciais de trabalhos relacionados à avaliação automatizada da idade óssea e à análise de radiografias pediátricas da mão.

### 2.1 The RSNA Pediatric Bone Age Machine Learning Challenge

**Referência:**  
HALABI, Safwan S. et al. The RSNA Pediatric Bone Age Machine Learning Challenge. *Radiology*, v. 290, n. 2, p. 498–503, 2019.

**Objetivo:**  
Apresentar o desafio da RSNA voltado à aplicação de técnicas de aprendizado de máquina para estimar a idade óssea a partir de radiografias pediátricas da mão.

**Dados utilizados:**  
O desafio disponibilizou 14.236 radiografias, sendo 12.611 para treinamento, 1.425 para validação e 200 para teste.

**Método:**  
Os participantes desenvolveram diferentes modelos de aprendizado de máquina para estimar a idade óssea. A principal métrica utilizada foi a Mean Absolute Distance (MAD), expressa em meses.

**Resultados:**  
Os cinco melhores resultados do desafio apresentaram MAD entre 4,2 e 4,5 meses.

**Relação com o projeto:**  
O artigo é diretamente relacionado ao trabalho porque apresenta o desafio e o conjunto de dados utilizados no projeto, além de permitir compreender a tarefa de regressão e a avaliação das estimativas de idade óssea.

### 2.2 Improving Automated Pediatric Bone Age Estimation Using Ensembles of Models from the 2017 RSNA Machine Learning Challenge

**Referência:**  
PAN, Ian et al. Improving Automated Pediatric Bone Age Estimation Using Ensembles of Models from the 2017 RSNA Machine Learning Challenge. *Radiology: Artificial Intelligence*, v. 1, n. 6, 2019.

**Objetivo:**  
Investigar formas de melhorar a estimativa automatizada da idade óssea utilizando modelos desenvolvidos a partir do desafio RSNA de 2017.

**Dados utilizados:**  
Foram utilizados os conjuntos do desafio, incluindo 12.611 radiografias de treinamento, 1.425 de validação e 200 imagens de teste.

**Método:**  
O trabalho analisou modelos desenvolvidos durante o desafio e investigou a utilização de conjuntos de modelos (ensembles) para melhorar a previsão da idade óssea.

**Resultados:**  
Os autores observaram que a combinação de previsões de diferentes modelos poderia melhorar o desempenho em relação à utilização de modelos individuais.

**Relação com o projeto:**  
O artigo é relevante por utilizar o mesmo problema e conjunto de dados do projeto. Entretanto, devido à restrição do trabalho contra métodos de deep learning, será utilizado principalmente como referência sobre o problema e a base de dados.

### 2.3 Automated Bone Age Assessment with Image Registration Using Hand X-ray Images

**Referência:**

ZULKIFLEY, Mohd Asyraf; ABDANI, Siti Raihanah; ZULKIFLEY, Nuraisyah Hani. Automated Bone Age Assessment with Image Registration Using Hand X-ray Images. *Applied Sciences*, v. 10, n. 20, p. 7233, 2020.

**Objetivo:**

Desenvolver um método automatizado para estimar a idade óssea a partir de radiografias da mão.

**Método:**

O trabalho utiliza técnicas de registro e padronização das imagens para reduzir diferenças de posição e orientação das radiografias. O método também emprega etapas de processamento da imagem e aumento de dados, utilizando diferentes rotações, fatores de zoom e cisalhamento para ampliar a variabilidade dos dados de treinamento.

**Resultados:**

O método proposto apresentou erro absoluto médio (MAE) de 8,200 meses e erro quadrático médio (MSE) de 121,902 meses².

**Relação com o projeto:**

O artigo é relevante para a etapa de pré-processamento, pois demonstra a importância da padronização das radiografias antes da estimativa da idade óssea. No presente projeto, embora seja utilizada uma abordagem mais simples, as imagens também passam por uma etapa de padronização antes da extração das características.


### 2.4 Autonomous Artificial Intelligence in Pediatric Radiology

**Referência:**  
THODBERG, Hans Henrik et al. Autonomous artificial intelligence in pediatric radiology: the use and perception of BoneXpert for bone age assessment. *Pediatric Radiology*, v. 52, p. 1338–1346, 2022.

**Objetivo:**  
Investigar a utilização do sistema automatizado BoneXpert na avaliação da idade óssea e verificar como os radiologistas utilizam a ferramenta na prática.

**Método:**  
Foi aplicado um questionário a radiologistas de departamentos europeus que utilizavam o software BoneXpert.

**Resultados:**  
O estudo identificou redução no tempo necessário para avaliação da idade óssea após a utilização do sistema. A maioria dos participantes utilizava a ferramenta como auxílio à avaliação, e não como substituição completa do radiologista.

**Relação com o projeto:**  
O estudo demonstra uma aplicação prática de sistemas automatizados na avaliação de idade óssea e contextualiza a utilização de ferramentas computacionais como apoio à análise de imagens médicas.

```python
import os
import random
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from PIL import Image

SEED = 42

random.seed(SEED)
np.random.seed(SEED)

BASE_DIR = "/kaggle/input"

TRAIN_CSV = os.path.join(
    BASE_DIR,
    "boneage-training-dataset.csv"
)

TRAIN_IMG_DIR = os.path.join(
    BASE_DIR,
    "boneage-training-dataset"
)

print("CSV encontrado:", os.path.exists(TRAIN_CSV))
print("Pasta de imagens encontrada:", os.path.exists(TRAIN_IMG_DIR))
```

There are 2 csv files in the current version of the dataset:


```python
df = pd.read_csv(TRAIN_CSV)

print("Quantidade de exames:", len(df))
print("Colunas:", df.columns.tolist())

display(df.head())
```

```python
N_AMOSTRA = 200

amostra = df.sample(
    n=N_AMOSTRA,
    random_state=SEED
).copy()

print("Quantidade de exames na amostra:", len(amostra))

display(amostra.head())
```

```python
registro = amostra.iloc[0]

exam_id = int(registro["id"])

print("ID do exame:", exam_id)
print("Idade óssea:", registro["boneage"], "meses")
print("Sexo (male):", registro["male"])
```

```python
img_path = os.path.join(
    TRAIN_IMG_DIR,
    f"{exam_id}.png"
)

print("Caminho da imagem:")
print(img_path)

print("\nImagem encontrada:", os.path.exists(img_path))
```

```python
import glob

padrao = os.path.join(
    TRAIN_IMG_DIR,
    f"*{exam_id}*"
)

arquivos_encontrados = glob.glob(padrao)

print("Arquivos encontrados:")
for arquivo in arquivos_encontrados[:10]:
    print(arquivo)

print("\nQuantidade encontrada:", len(arquivos_encontrados))
```

```python
import os

print("Conteúdo de /kaggle/input/boneage-training-dataset:")

itens = os.listdir(TRAIN_IMG_DIR)

print("Quantidade de itens:", len(itens))

for item in itens[:20]:
    print(item)
```

```python
REAL_IMG_DIR = os.path.join(
    TRAIN_IMG_DIR,
    "boneage-training-dataset"
)

print("Pasta real:", REAL_IMG_DIR)
print("Existe:", os.path.exists(REAL_IMG_DIR))

print("\nPrimeiros arquivos:")
for arquivo in os.listdir(REAL_IMG_DIR)[:10]:
    print(arquivo)
```

```python
img_path = os.path.join(
    REAL_IMG_DIR,
    f"{exam_id}.png"
)

print("Imagem:", img_path)
print("Encontrada:", os.path.exists(img_path))
```

```python
img = Image.open(img_path)

print("Formato da imagem:", img.format)
print("Tamanho:", img.size)
print("Modo:", img.mode)

display(img)
```

```python
arr = np.array(img)

print("Shape:", arr.shape)
print("Valor mínimo:", arr.min())
print("Valor máximo:", arr.max())
print("Média dos pixels:", arr.mean())
print("Desvio padrão:", arr.std())
```

```python
plt.figure(figsize=(8, 4))

plt.hist(arr.ravel(), bins=256)

plt.title("Distribuição das intensidades dos pixels")
plt.xlabel("Intensidade do pixel")
plt.ylabel("Quantidade de pixels")

plt.show()
```

```python
print("=== INFORMAÇÕES DO DATASET ===")

print("Quantidade de exames:", len(df))

print("\nDistribuição do sexo:")
print(df["male"].value_counts())

print("\nDistribuição do sexo (%):")
print(df["male"].value_counts(normalize=True) * 100)

print("\nIdade óssea:")
print("Mínima:", df["boneage"].min(), "meses")
print("Máxima:", df["boneage"].max(), "meses")
print("Média:", df["boneage"].mean(), "meses")
print("Mediana:", df["boneage"].median(), "meses")
print("Desvio padrão:", df["boneage"].std(), "meses")
```

```python
plt.figure(figsize=(9, 5))

plt.hist(df["boneage"], bins=30)

plt.title("Distribuição da idade óssea")
plt.xlabel("Idade óssea (meses)")
plt.ylabel("Quantidade de exames")

plt.show()
```

```python
faixas = [
    (0, 60),
    (61, 120),
    (121, 180),
    (181, 228)
]

fig, axes = plt.subplots(1, 4, figsize=(16, 5))

for ax, (idade_min, idade_max) in zip(axes, faixas):

    grupo = df[
        (df["boneage"] >= idade_min) &
        (df["boneage"] <= idade_max)
    ]

    registro = grupo.sample(n=1, random_state=SEED).iloc[0]

    caminho = os.path.join(
        REAL_IMG_DIR,
        f"{int(registro['id'])}.png"
    )

    imagem = Image.open(caminho)

    ax.imshow(imagem, cmap="gray")
    ax.set_title(
        f"{idade_min}–{idade_max} meses\n"
        f"Idade: {int(registro['boneage'])} meses"
    )
    ax.axis("off")

plt.tight_layout()
plt.show()
```
